###  Dataset and Analysis Scripts Overview

In an effort to promote open science, all the datasets of this project are hosted in Open Science Framework (OSF) at DOI: https://doi.org/10.17605/OSF.IO/STK2R. Similarly, all the analyses, going from raw data to figure panels and statistical tests, are hosted in a GitHub repository at https://github.com/AlexLM96/Legaria_etal_2025/tree/main. 

In addition to the availability of both raw data and analyses scripts, we provide tools for easily engaging with the data interactively. This allows for the running analyses scripts and customization of the analyses scripts on a browser, without having to worry about any environment or dependency conflicts. We achieve this using two independent tools: Google Colab and BinderHub.




### Using Google Colab to view and customize analyses scripts

Using Google Colab (heron "Colab") for this project is very simple. There is an ipython notebook (.ipynb) for every figure (Figure 1 has 3 independent notebooks) in the GitHub Repository. On the README repository page, you can click the google Colab badge to each figure. By default, usersr cannot run or modify the code. This can be done in two ways:

1. **Activate "Playground mode" (preferred method).** Go to File -> Open in playground mode. This enables running and customizing the notebooks. None of the changes made will be saved to the original file nor modify it in any way, instead all changes will be discarded when the tab is closed.

2. **Create a copy of the file.** Go to to File -> Save a copy to Drive. This will create a copy of the notebook in your Google Drive, which can be run and customized. All changes made will be saved to this copy.

Importantly, to use the Google Colab option, a Google account is needed.

### Using MyBinder to view and customize analyses scripts

Project Binder is an open community that makes it possible to create sharable, interactive, reproducible environments. In simple terms, it is a combination of Docker and cloud-based JupyterLab. 

Clicking on the binder badge in the README page of the project's GitHub repository creates a binder instance (Docker container) hosted by JupyterHub. In other words, you will be able to navigate, run and customize all the notebooks using the exact same environment that was used for creating these notebooks. None of the changes made will be saved to the original file nor modify it in any way, instead all changes will be discarded when the tab is closed.
